```python
# calcula a média, a maior e a menor nota de uma lista de 4 notas
notas = [7.5, 8.0, 6.5, 9.0]
media = sum(notas) / len(notas)
maior = max(notas)
menor = min(notas)

print(f"Média final: {media:.2f}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")
```

```python
# verifica aprovação com base na média
notas = [6.0, 7.5, 5.5, 8.0]
media = sum(notas) / len(notas)

if media >= 7:
    print("APROVADO")
else:
    nota_final = 6.0  # exemplo de nota da prova final
    nova_media = (media + nota_final) / 2
    if nova_media >= 5:
        print("APROVADO")
    else:
        print("REPROVADO")
```

```python
# verifica se um número é par ou ímpar usando função lambda
numero = int(input("Digite um número: "))
verifica_par_impar = lambda x: "Par" if x % 2 == 0 else "Ímpar"
print(verifica_par_impar(numero))
```

```python
# realiza uma operação aritmética simples entre dois números com base na escolha do usuário
operacoes = {
    "soma": lambda x, y: x + y,
    "subtracao": lambda x, y: x - y,
    "multiplicacao": lambda x, y: x * y,
    "divisao": lambda x, y: x / y if y != 0 else "Divisão por zero não permitida"
}

operacao = input("Qual operação deseja? (soma, subtracao, multiplicacao, divisao): ").strip().lower()
x = float(input("Digite o primeiro número: "))
y = float(input("Digite o segundo número: "))

resultado = operacoes.get(operacao, lambda a, b: "Operação inválida")(x, y)
print("Resultado:", resultado)
```

```python
# converte uma lista de temperaturas em Celsius para Fahrenheit
celsius = [22.5, 40, 13, 29, 34]
fahrenheit = list(map(lambda c: round((c * 9/5) + 32, 2), celsius))
print("Temperaturas em Fahrenheit:", fahrenheit)
```

```python
# calcula a soma dos valores pares e dos valores ímpares de uma lista
numeros = [21, 5, 34, 8, 16, 7, 3]
pares = list(filter(lambda x: x % 2 == 0, numeros))
impares = list(filter(lambda x: x % 2 != 0, numeros))
print("Soma dos pares:", sum(pares))
print("Soma dos ímpares:", sum(impares))
```

```python
# encontra o maior e o menor valor da sequência
from functools import reduce

numeros = [54, 10, 29, 87, 7, 64]

# maior valor
maior = reduce(lambda x, y: x if x > y else y, numeros)
# menor valor
menor = reduce(lambda x, y: x if x < y else y, numeros)

print("Maior valor:", maior)
print("Menor valor:", menor)
```

```python
# jogo acerte o número
import random

def jogar_novamente():
    """Pergunta ao usuário se ele deseja jogar novamente."""
    resposta = input("Deseja jogar novamente? (s/n): ").lower()
    return resposta == 's'

def validar_input(input_usuario):
    """Verifica se o input do usuário é um número inteiro válido dentro do intervalo 0-10."""
    try:
        numero = int(input_usuario)
        if 0 <= numero <= 10:
            return numero
        else:
            print("Por favor, insira um número entre 0 e 10.")
            return None
    except ValueError:
        print("Entrada inválida! Por favor, insira um número inteiro.")
        return None

def jogo_acertar_numero():
    """Função principal do jogo, onde o computador escolhe o número e o jogador tenta adivinhar."""
    print("Bem-vindo ao jogo! Eu escolhi um número entre 0 e 10.")

    numero_computador = random.randint(0, 10)
    tentativas = 5
    pontos = 0

    for tentativa in range(1, tentativas + 1):
        print(f"Tentativa {tentativa} de {tentativas}:")

        while True:
            palpite = input(f"Qual o seu palpite? (0 a 10): ")
            numero_usuario = validar_input(palpite)
            if numero_usuario is not None:
                break

        if numero_usuario == numero_computador:
            print(f"Parabéns! Você acertou o número {numero_computador}!")
            if tentativa == 1:
                pontos = 100
            else:
                pontos = max(10, 100 - 10 * (tentativa - 1))
            break
        else:
            print(f"Errado! O número era {numero_computador}. Tente novamente.")

    if pontos == 0:
        print(f"Você não acertou o número. O número era {numero_computador}.")

    print(f"Sua pontuação final: {pontos} pontos.")

def main():
    """Função principal para controlar o fluxo do jogo."""
    while True:
        jogo_acertar_numero()
        if not jogar_novamente():
            print("Obrigado por jogar! Até logo!")
            break

main()
```
