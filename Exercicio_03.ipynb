```python
# encontra números que são divisíveis por 7, mas não são múltiplos de 5
def find_numbers(start, stop):
    result = [str(i) for i in range(start, stop + 1) if i % 7 == 0 and i % 5 != 0]
    print(",".join(result))

find_numbers(100, 200)
```

```python
# gera o quadrado de 1 até n
def generate_square_dict(n):
    return {i: i**2 for i in range(1, n + 1)}

print(generate_square_dict(8))
```

```python
# gera uma lista e uma tupla de uma sequência de números
def convert_input_to_list_and_tuple(data):
    items = data.split(",")
    return items, tuple(items)

print(convert_input_to_list_and_tuple("3,6,5,3,2,8"))
```

```python
# transforma uma string em letras maiúsculas
class InputOutString:
    def __init__(self):
        self.text = ""

    def get_string(self):
        self.text = input("Digite uma string: ")

    def print_string(self):
        print(self.text.upper())

# teste com a string 'hello world'
str_obj = InputOutString()
str_obj.get_string()
str_obj.print_string()
```

```python
# calcula os valores possíveis de Q
import math

def calculate_q_values(d_values):
    C, H = 50, 30
    result = [str(round(math.sqrt((2 * C * float(d)) / H))) for d in d_values.split(",")]
    print(",".join(result))

calculate_q_values("100,150,180")
```

```python
# gera um array bidimensional a partir de 2 dígitos
def create_matrix(m, n):
    matrix = [[i * j for j in range(n)] for i in range(m)]
    for row in matrix:
        print(row)

create_matrix(4, 3)
```

```python
# ordena uma lista de palavras (separadas por vírgulas) por ordem alfabética
def sort_words(words):
    sorted_list = sorted(words.split(','))
    return ",".join(sorted_list)

print(sort_words('banana,apple,grape,orange'))
```

```python
# transforma frases em maiúsculas
def capitalize_lines(lines):
    return [[line.upper() for line in lines]]

print(capitalize_lines(['Sundays are Fun', 'Monday is Done']))
```

```python
# ordena uma lista de palavras (separadas por espaços) por ordem alfabética
def unique_sorted_words(text):
    words = text.split()
    unique_sorted = sorted(set(words))
    return " ".join(unique_sorted)

print(unique_sorted_words('dog cat apple cat banana dog'))
```

```python
# verifica números divisíveis por 5 em binário
def binary_divisible_by_5(binaries):
    result = [b for b in binaries.split(',') if int(b, 2) % 5 == 0]
    return ",".join(result)

print(binary_divisible_by_5('1101,1010,1111,1001'))
```

```python
# retorna todos os números cujos dígitos são pares entre 1200 e 2100
def even_digit_numbers(start, end):
    result = [str(num) for num in range(start, end + 1) if all(int(d) % 2 == 0 for d in str(num))]
    return ",".join(result)

print(even_digit_numbers(1200, 2010))
```

```python
# calcula o número de letras e dígitos de uma frase
def count_letters_digits(text):
    letters = sum(c.isalpha() for c in text)
    digits = sum(c.isdigit() for c in text)
    return {"LETTERS": letters, "DIGITS": digits}

print(count_letters_digits('Data123 Science 2024'))
```

```python
# verifica quantas letras maiúsculas e minúsculas têm em uma frase
def count_case(text):
    upper = sum(c.isupper() for c in text)
    lower = sum(c.islower() for c in text)
    return {"UPPER CASE": upper, "LOWER CASE": lower}

print(count_case('Hello World!'))
```

```python
# calcula o valor de a + aa + aaa + aaaa para um determinado dígito a
def sum_of_series(a):
    a = str(a)
    total = int(a) + int(a*2) + int(a*3) + int(a*4)
    return total

print(sum_of_series(7))
```

```python
# calcula a área do círculo com base no valor do raio
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return round(3.1416 * self.radius ** 2, 2)

aCircle = Circle(3)
print(aCircle.area())
```

```python
# calcula o quadrado de um número
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, length):
        self.length = length

    def area(self):
        return self.length ** 2

aSquare = Square(4)
print(aSquare.area())
```

```python
# calcula e captura o erro
def catch_zero_division():
    try:
        10 / 0
    except ZeroDivisionError as e:
        return str(e)

print(catch_zero_division())
```

```python
# captura o nome de usuário de um e-mail
import re

def extract_username(email):
    match = re.match(r"([^@]+)@.+", email)
    return match.group(1) if match else ""

print(extract_username("alice@company.com"))
```

```python
# calcula o valor de n
def f(n):
    if n == 0:
        return 1
    return f(n - 1) + 100

print(f(5))
```

```python
# gera 5 números aleatórios entre 150 e 250
import random

def generate_random_numbers():
    return random.sample(range(150, 251), 5)

print(generate_random_numbers())
```

```python
# gera 5 números pares aleatórios entre 150 e 250
import random

def generate_even_numbers():
    evens = list(range(150, 251, 2))
    return random.sample(evens, 5)

print(generate_even_numbers())
```
